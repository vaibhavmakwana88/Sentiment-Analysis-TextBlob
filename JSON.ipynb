## Step 3: Convert JSON Dictionary into DataFrame

We transformed the raw `tweets` dictionary into a structured pandas DataFrame.  
Now each tweet is represented as a row with proper columns:

- **USERIDs** → unique identifier for each tweet  
- **tweet_author** → the person or organization who posted the tweet  
- **tweet_text** → the actual content of the tweet  

This structure makes it easier to apply sentiment analysis and visualization

```python
df=pd.DataFrame.from_dict(tweets,orient="index")
df=df.reset_index().rename(columns={"index":"USERIDs"})
```

USING REGEX LIBRARY TO CLEAN EXTRA SYMBOLS AND EMOJIS

```python
import re

```

```python
def clean_text(text):
    text= re.sub(r'[^\w\s]',"",text)
    return(text)
```

```python
df3=df["tweet_text"].apply(clean_text)
```

Getting polarity and subjectivity score

```python
def sentiment_polarity(text):
    score= TextBlob(text).sentiment.polarity
    return score
```

```python
def sentiment_subjectivity(text):
    score= TextBlob(text).sentiment.subjectivity
    return score
```

```python
df3["polarity_score"]=df3["tweet_text"].apply(sentiment_polarity)
```

```python
df3["subjectivity_score"]=df3["tweet_text"].apply(sentiment_subjectivity)
```

For easy and readable results


```python
def polarity_meaning(score):
    if score>0:
        return "Positive"
    elif score<0:
        return "Negative"
    else:
        return "Neutral"

def subjectivity_meaning(score):
    if score>=0.5:
        return "Subjective"
    
    else:
        return "Objective"
```

```python
df4["polarity_meaning"]=df3["polarity_score"].apply(polarity_meaning)
```

```python
df3["subjectivity_meaning"]=df3["polarity_score"].apply(subjectivity_meaning)
```
